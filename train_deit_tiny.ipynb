## Imports

```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)

import torch
import os

from src.config import MODEL_CONFIGS
from src.data import get_dataloaders
from src.model import create_model, print_model_params
from src.train import train_model
from src.visualize import plot_learning_curves

cfg = MODEL_CONFIGS['deit_tiny']
```

## Device Setup & Reproducibility

```python
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")
print(f"Using device: {device}")

torch.manual_seed(42)
if device.type == 'cuda':
    torch.cuda.manual_seed(42)

os.makedirs(cfg.output_dir, exist_ok=True)
```

## Data Preparation

```python
train_loader, test_loader = get_dataloaders(batch_size=32)
print(f"Train: {len(train_loader.dataset)} images | Test: {len(test_loader.dataset)} images")
```

## Model Setup

```python
model = create_model(cfg, device)
print_model_params(model)
```

## Training & Validation

```python
history, best_acc = train_model(cfg, model, train_loader, test_loader, device, num_epochs=20)
```

## Learning Curves

```python
plot_learning_curves(history, cfg.output_dir)
```
