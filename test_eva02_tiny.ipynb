## Imports

```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)

import torch
import os

from src.config import MODEL_CONFIGS
from src.data import get_test_loader
from src.model import create_model, print_model_params
from src.evaluate import predict, save_predictions_csv, compute_and_print_metrics
from src.visualize import plot_samples, plot_confusion_matrix, plot_attention_maps
from src.inference import measure_inference_time

cfg = MODEL_CONFIGS['eva02_tiny']
```

## Device Setup

```python
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")
print(f"Using device: {device}")

torch.manual_seed(42)
if device.type == 'cuda':
    torch.cuda.manual_seed(42)

os.makedirs(cfg.output_dir, exist_ok=True)
```

## Data Preparation

```python
test_loader, test_dataset = get_test_loader(batch_size=32)
print(f"Test: {len(test_dataset)} images")
```

## Model Setup

```python
model = create_model(cfg, device)
print_model_params(model)
```

## Performance Metrics

```python
all_images, all_preds, all_labels = predict(cfg, model, test_loader, device)
```

```python
plot_samples(cfg, all_images, all_labels, all_preds)
```

```python
save_predictions_csv(cfg, all_labels, all_preds)
```

```python
compute_and_print_metrics(all_labels, all_preds)
```

## Confusion Matrix

```python
plot_confusion_matrix(cfg, all_labels, all_preds)
```

## Inference Timing

```python
measure_inference_time(model, test_loader, device)
```

## Attention Maps

```python
plot_attention_maps(cfg, model, test_dataset, device)
```
